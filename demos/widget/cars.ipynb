# A Great Tables table as a notebook widget

Run each cell in turn (with Shift+Enter). Select a few cars, or filter by manufacturer, then run the cells below the table. See [GTI in Python and Apps](https://posit-dev.github.io/great-tables/user-guide/gti-in-python-and-apps.html).

```python
from great_tables import GT, GTI, data

cars = data.gtcars[["mfr", "model", "year", "hp", "msrp"]]
widget = (
    GTI(cars)
    .fmt_integer("hp")
    .fmt_currency("msrp", decimals=0)
    .i_cols_filter(columns="mfr", type="facet")
    .i_selection("multiple")
    .i_as_widget()
)
widget
```

```python
chosen = widget.view(selected=True)
print(f"{len(chosen)} of {len(widget.rows)} cars selected")
GT(chosen).fmt_integer("hp").fmt_currency("msrp", decimals=0)
```

```python
widget.query = 'mfr in ("Ferrari", "Lamborghini")'
widget.sort = "-hp"
widget.selection = []
```
